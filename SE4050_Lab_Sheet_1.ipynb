```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from mpl_toolkits.mplot3d import Axes3D  # for 3D plotting

# Set style for better plots
sns.set_theme(style='whitegrid')
plt.rcParams['figure.dpi'] = 100
```

```python
exp_4x4 = np.random.exponential(scale=1.0, size=(4,4))
print("Random 4x4 array from exponential distribution (scale=1):")
print(exp_4x4)
```

```python
# Generate samples
n = 100000
exp_samples = np.random.exponential(scale=1.0, size=n)
uni_samples = np.random.uniform(0, 1, size=n)
norm_samples = np.random.normal(0, 1, size=n)

# Plot histograms
plt.figure(figsize=(10, 6))
plt.hist(exp_samples, density=True, bins=80, histtype='step', color='red', label='Exponential')
plt.hist(uni_samples, density=True, bins=80, histtype='step', color='blue', label='Uniform')
plt.hist(norm_samples, density=True, bins=80, histtype='step', color='green', label='Normal')

# Adjust axis limits and labels
plt.xlim(-2, 6)
plt.ylim(0, 1.2)
plt.xlabel('Value')
plt.ylabel('Density')
plt.title('Comparison of Exponential, Uniform, and Normal Distributions')
plt.legend()
plt.show()
```

```python
x = np.linspace(-5, 5, 50)
y = np.linspace(-5, 5, 50)
X, Y = np.meshgrid(x, y)
Z = X**2 + Y**2

fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')
surf = ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('Surface plot of Z = X² + Y²')
fig.colorbar(surf, shrink=0.5, aspect=10)
plt.show()
```

```python
# Load Pokémon stats dataset
url = "https://raw.githubusercontent.com/KeithGalli/pandas/master/pokemon_data.csv"
pokemon = pd.read_csv(url)

# The columns we need are exactly these names:
features = ['HP', 'Attack', 'Defense', 'Sp. Atk', 'Sp. Def', 'Speed']

# Extract the stats dataframe
stats_df = pokemon[features].copy()

# Compute correlations
pearson_corr = stats_df.corr(method='pearson')
spearman_corr = stats_df.corr(method='spearman')

# Plot heatmaps
fig, axes = plt.subplots(1, 2, figsize=(14, 6))
sns.heatmap(pearson_corr, annot=True, fmt='.2f', cmap='coolwarm', square=True, ax=axes[0])
axes[0].set_title('Pearson Correlation')
sns.heatmap(spearman_corr, annot=True, fmt='.2f', cmap='coolwarm', square=True, ax=axes[1])
axes[1].set_title('Spearman Rank Correlation')
plt.tight_layout()
plt.show()
```

```python
from google.colab import files

# Upload the CSV file
uploaded = files.upload()

filename = list(uploaded.keys())[0]
print(f"Reading file: {filename}")

# Load the dataset
df_au = pd.read_csv(filename)

# Display column names to verify
print("Columns in the dataset:")
print(df_au.columns.tolist())
```

```python
# Keep only the specified columns
selected_cols = ['N_total', 'N_bulk', 'N_surface', 'R_avg']
new_df = df_au[selected_cols].copy()
new_df.head()
```

```python
print("First 20 samples of the filtered dataframe:")
print(new_df.head(20))
```

```python
stats_au = new_df.describe(percentiles=[0.25, 0.5, 0.75])
stats_au.loc['mean'] = new_df.mean()
stats_au.loc['std'] = new_df.std()
print("Summary statistics for the four features:")
print(stats_au)
```

```python
fig, axes = plt.subplots(1, 4, figsize=(16, 4))
for i, col in enumerate(selected_cols):
    axes[i].hist(new_df[col], bins=30, edgecolor='black', alpha=0.7)
    axes[i].set_title(col)
    axes[i].set_xlabel('Value')
    axes[i].set_ylabel('Frequency')
plt.tight_layout()
plt.show()
```
